```python
# informa o caminho do shape
shape_path = "/content/drive/MyDrive/shapes/bairros.shp"

# associa o shape à variavel
gdf = gpd.read_file(shape_path, encoding='utf-8')
```

```python
#exibe as informações armazenadas apenas na coluna NOME
gdf.NOME
```

```python
#desenha o shape ordenado pela coluna nome
gdf.plot("NOME", cmap="RdYlGn")
```

```python
#desenha o shape ordenado pela coluna nome e adiciona legenda
ax = gdf.plot(figsize=(30, 25))
gdf.plot(ax=ax, column='NOME', categorical=True,
             legend=True, cmap='RdYlGn')
```

```python
#retorna informações dos parametros geodésicos
gdf.crs
```

```python
#cria uma nova variavel: porto alegre_utm
#seleciona o shape da variavel gdf
#altera o os parametros geodesicos
#salva o resultado na variavel recem criada
gdf_utm = (gdf.to_crs('EPSG:32722'))
```

```python
#retorna informações dos parametros geodésicos
gdf_utm.crs
```

```python
#calcula a área dos polígonos
gdf_utm['area_m2'] = gdf_utm['geometry'].area.astype('int64')
```

```python
gdf_utm['length_m'] = gdf_utm['geometry'].length.astype('int64')
```

```python
gdf_utm
```

```python
ax = gdf_utm.plot(figsize=(22,10))
ax.set_title(gdf_utm.crs.name)
```

```python
out='/content/drive/MyDrive/shapes/porto_alegre_utm.shp'
gdf_utm.to_file(out)
```

```python
# importa a biblioteca folium
import folium
#cria a variavel map e associa o mapa "open street map" da biblioteca folium
map = folium.Map(location=[-30.1,-51.15], zoom_start = 15)
#manda desenhar o map
map
```

```python
#chama o shape armazenado na varivel gdf e adiciona ao map
folium.GeoJson(data=gdf["geometry"]).add_to(map)
#manda desenhar o map
map
```

```python
#CONSULTA A TABELA DE ATRIBUTOS
# criar a variavel restiga
# a variavel vai armazenar o resultado da consulta
#PARA FAZER A CONSULTA, CHAMAR A VARIAVEL ONDE ESTÁ ARMAZENADO O SHAPE "gdf"
#a consulta é NOME == RESTINGA
restinga = gdf[gdf.NOME=='RESTINGA' ]
#chama a nova variavel, que vai ser mostrada na saida
restinga
```

```python
#plota o vetor, resultado da consulta
restinga.plot()
```

```python
#cria um caminho para salvar a saida
#manda a consulta para um arquivo shape

out='/content/drive/MyDrive/shapes/restinga_wgs84.shp'
restinga.to_file(out)
```

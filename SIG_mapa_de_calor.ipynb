```python
# Importar as bibliotecas
import geopandas as gpd
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# Definir o caminho para os dados a serem lidos
# a pasta no drive e o nome do shape
# os dados tem que estar no formato shapefile
points_shape = r"/content/drive/MyDrive/shapes/pontos_shape.shp"

# Ler os dados e armazena na variável points que é do tipo geodataframe
points = gpd.read_file(points_shape, encoding='utf-8')
```

```python
# Verifica o SRC do geodataframe
points.crs
```

```python
points
```

```python
# Mostra as primeiras 5 linhas do geodataframe
points.head()
```

```python
# desenha os pontos
ax = points.plot(markersize = 10, color = "red")
```

```python
# Importar a biblioteca folium para criar mapas
import folium

# https://python-visualization.github.io/folium/latest/user_guide.html

# Define a centralização inicial do mapa (location=[lat,long]), nível de zoom (zoom=), visibilidade da escala (control_scale=)
# Trava os controles do zoom (zoom_control=False, scrollWheelZoom=False, dragging=False)


m = folium.Map(location=[-30.1, -51.2], zoom_start=12, zoom_control=False, scrollWheelZoom=False, dragging=False, control_scale=True)
m
```

```python
# Importar o plugin MousePosition para retornar a coordenada com a navegação do mouse
from folium.plugins import MousePosition

# adiciona o plugin ao mapa base
MousePosition().add_to(m)
```

```python
# Visualize o mapa base (sem os pontos)
m
```

```python

# Iterar pelas linhas do geodataframe e adicionar um marcador com popup para cada ponto
for index, row in points.iterrows():
    folium.Marker(
        location=[row['lat'], row['long']],
        popup=row['nome']  # Usa a coluna 'nome' para o popup
    ).add_to(m)

# Visualize o mapa com os marcadores e popups
m
```

```python
# Obter as coordenadas x e y de cada ponto e cria uma lista de pares (tuplas) a partir do geodataframe
points["x"] = points["geometry"].apply(lambda geom: geom.x)
points["y"] = points["geometry"].apply(lambda geom: geom.y)

# armazena os pares na variavel coordenadas
coordenadas = list(zip(points["y"], points["x"]))
```

```python
# mostra a lista de pares (lat / long)
coordenadas
```

```python
# Importar o plugin que gera os mapas de calor
from folium.plugins import HeatMap
```

```python
# Adicionar o mapa de calor ao mapa base
# é feito com base na lista de coordenadas (locations) e não do geodataframe diretamente (points)

HeatMap(coordenadas).add_to(m)

# Visualize o resultado
m
```

```python
# Cria um novo mapa base, sem os pontos, so com o mapa de calor
m_calor = folium.Map(location=[-30.1, -51.2], zoom_start=12, zoom_control=False, scrollWheelZoom=True, dragging=True, control_scale=True)

HeatMap(coordenadas).add_to(m_calor)
m_calor
```

```python
# Importar o plugin de clusters
from folium.plugins import MarkerCluster
```

```python
# Cria um marcador do tipo cluster a partir das localizações dos pontos
marker_cluster = MarkerCluster(coordenadas)

# Altera os controles de zoom do mapa base
m = folium.Map(location=[-30.1, -51.2], zoom_start=11, zoom_control=True, scrollWheelZoom=True, dragging=True, control_scale=True)
#m = folium.Map(location=[-30.1, -51.2], zoom_start=11, tiles="CartoDB positron", zoom_control=True, scrollWheelZoom=True, dragging=True, control_scale=True)
# Adicione o marcador ao mapa base
marker_cluster.add_to(m)

# Visualize o mapa com os clusters
m
```

```python
# Iterar pela lista coordenadas e adicionar um marcador para cada par de coordenadas
for lat, lng in coordenadas:
   folium.Marker([lat, lng], icon=folium.Icon(color='green')).add_to(m) # Define a cor do marcador como verde
# Visualize the map with markers

m
```
